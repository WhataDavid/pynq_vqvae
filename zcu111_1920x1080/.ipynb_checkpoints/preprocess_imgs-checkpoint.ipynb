```python
# import os
# import numpy as np
# from PIL import Image as PILImage

# 版本一
# # --- 配置 ---
# IMG_DIR = '/home/xilinx/jupyter_notebooks/duxu/pynq_vqvae/imgs'
# SAVE_DIR = '/home/xilinx/jupyter_notebooks/duxu/pynq_vqvae/imgs_preprocessed_1920x1080'
# TARGET_SIZE = (1920, 1080)  # (W, H)
# # 这里的 scale 需要和你的 DPU Encoder 输入 scale 一致 (2^-6 = 0.015625)
# ENC_IN_SCALE = 0.015625 

# if not os.path.exists(SAVE_DIR):
#     os.makedirs(SAVE_DIR)

# files = sorted([f for f in os.listdir(IMG_DIR) if f.lower().endswith(('.png', '.jpg', '.jpeg'))])

# print(f"开始预处理 {len(files)} 张图片...")

# for f in files:
#     # 1. 加载与缩放 (与训练/原始逻辑保持一致)
#     img_pil = PILImage.open(os.path.join(IMG_DIR, f)).convert('RGB')
#     img_small = img_pil.resize(TARGET_SIZE, PILImage.BILINEAR)
    
#     # 2. 归一化到 [-1, 1]
#     img_arr = np.asarray(img_small, dtype=np.float32)
#     input_fp32 = (img_arr / 255.0 - 0.5) / 0.5
    
#     # 3. 量化为 int8 (DPU 直接输入格式)
#     input_int8 = np.clip(np.round(input_fp32 / ENC_IN_SCALE), -128, 127).astype(np.int8)
    
#     # 4. 保存为 numpy 格式，读取速度极快
#     save_path = os.path.join(SAVE_DIR, f + '.npy')
#     np.save(save_path, input_int8)

# print(f"✅ 处理完成，数据保存在: {SAVE_DIR}")

#!/usr/bin/env python3
# -*- coding: utf-8 -*-

# 版本二
# import os
# import numpy as np
# import cv2

# # ============================================================
# # 配置
# # ============================================================
# IMG_DIR = '/home/xilinx/jupyter_notebooks/duxu/pynq_vqvae/imgs'
# SAVE_DIR = '/home/xilinx/jupyter_notebooks/duxu/pynq_vqvae/imgs_preprocessed_1920x1080'

# TARGET_W = 1920
# TARGET_H = 1080

# # Encoder input fix_point=6 -> scale=2^-6=0.015625
# ENC_IN_SCALE = 0.015625

# os.makedirs(SAVE_DIR, exist_ok=True)

# exts = ('.png', '.jpg', '.jpeg', '.bmp')
# files = sorted([
#     f for f in os.listdir(IMG_DIR)
#     if f.lower().endswith(exts)
# ])

# print("=" * 80)
# print("Preprocess images with OpenCV, aligned with PYNQ Encoder input")
# print("=" * 80)
# print("IMG_DIR :", IMG_DIR)
# print("SAVE_DIR:", SAVE_DIR)
# print("TARGET  :", TARGET_W, TARGET_H)
# print("scale   :", ENC_IN_SCALE)
# print("files   :", len(files))

# if len(files) == 0:
#     raise RuntimeError(f"No image files found in {IMG_DIR}")

# for i, f in enumerate(files):
#     in_path = os.path.join(IMG_DIR, f)

#     # 1. OpenCV 读取：BGR
#     img_bgr = cv2.imread(in_path)
#     if img_bgr is None:
#         raise FileNotFoundError(f"Cannot read image: {in_path}")

#     # 2. BGR -> RGB
#     img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

#     # 3. resize，与 PYNQ test_cfg_1920x1080.py 对齐
#     img_resized = cv2.resize(
#         img_rgb,
#         (TARGET_W, TARGET_H),
#         interpolation=cv2.INTER_LINEAR
#     )

#     # 4. normalize 到 [-1, 1]
#     input_fp32 = img_resized.astype(np.float32)
#     input_fp32 = (input_fp32 / 255.0 - 0.5) / 0.5

#     # 5. quantize 到 int8
#     input_int8 = np.clip(
#         np.round(input_fp32 / ENC_IN_SCALE),
#         -128,
#         127
#     ).astype(np.int8)

#     # 6. 保存 HWC int8，shape=(1080,1920,3)
#     save_path = os.path.join(SAVE_DIR, f + '.npy')
#     np.save(save_path, input_int8)

#     if i < 5 or (i + 1) % 20 == 0 or (i + 1) == len(files):
#         print(
#             f"[{i+1:04d}/{len(files):04d}] {f} -> {os.path.basename(save_path)} "
#             f"shape={input_int8.shape}, "
#             f"range=[{int(input_int8.min())},{int(input_int8.max())}], "
#             f"mean={float(input_int8.mean()):.3f}, std={float(input_int8.std()):.3f}",
#             flush=True
#         )

# print("=" * 80)
# print("✅ 处理完成")
# print("数据保存在:", SAVE_DIR)
# print("=" * 80)

# 版本三
#!/usr/bin/env python3
# -*- coding: utf-8 -*-

import os
import numpy as np
import torch
import torchvision
import torchvision.transforms as transforms
from PIL import Image as PILImage

# ============================================================
# 配置
# ============================================================
IMG_DIR = '/home/xilinx/jupyter_notebooks/duxu/pynq_vqvae/imgs'
SAVE_DIR = '/home/xilinx/jupyter_notebooks/duxu/pynq_vqvae/imgs_preprocessed_1920x1080'

TARGET_H = 1080
TARGET_W = 1920

# Encoder input fix_point=6 -> scale=2^-6=0.015625
ENC_IN_SCALE = 0.015625

os.makedirs(SAVE_DIR, exist_ok=True)

exts = ('.png', '.jpg', '.jpeg', '.bmp')
files = sorted([
    f for f in os.listdir(IMG_DIR)
    if f.lower().endswith(exts)
])

print("=" * 80)
print("Preprocess images with torchvision, aligned with Windows pipeline")
print("=" * 80)
print("IMG_DIR :", IMG_DIR)
print("SAVE_DIR:", SAVE_DIR)
print("TARGET H,W:", TARGET_H, TARGET_W)
print("scale   :", ENC_IN_SCALE)
print("files   :", len(files))
print("torch version      :", torch.__version__)
print("torchvision version:", torchvision.__version__)
print("=" * 80)

if len(files) == 0:
    raise RuntimeError(f"No image files found in {IMG_DIR}")

# ============================================================
# Windows 端等价流程:
#   transforms.ToTensor()
#   transforms.Resize((TARGET_H, TARGET_W))
#   transforms.Normalize((0.5,0.5,0.5),(0.5,0.5,0.5))
#
# 注意 Resize 参数顺序是 (H, W)，不是 (W, H)
# ============================================================

# 如果你的 Windows torchvision 版本较新，Resize 默认 antialias=True。
# 为了尽量稳定，这里显式设置 antialias=True。
# 如果你的 Windows 版本很老，不支持 antialias 参数，把 antialias=True 删除即可。
try:
    preprocess = transforms.Compose([
        transforms.ToTensor(),  # PIL RGB -> torch float32, CHW, [0,1]
        transforms.Resize((TARGET_H, TARGET_W), antialias=True),
        transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5)),
    ])
except TypeError:
    print("WARNING: torchvision Resize does not support antialias argument; using default Resize.")
    preprocess = transforms.Compose([
        transforms.ToTensor(),
        transforms.Resize((TARGET_H, TARGET_W)),
        transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5)),
    ])

for i, f in enumerate(files):
    in_path = os.path.join(IMG_DIR, f)

    # 1. PIL 读取 RGB，和 torchvision ImageFolder 默认 PIL 路线一致
    img = PILImage.open(in_path).convert('RGB')

    # 2. ToTensor -> Resize -> Normalize
    x = preprocess(img)  # shape: (3, 1080, 1920), float32, range roughly [-1,1]

    # 3. CHW -> HWC
    x_hwc = x.permute(1, 2, 0).contiguous().numpy().astype(np.float32)

    # 4. quantize 到 int8
    input_int8 = np.clip(
        np.round(x_hwc / ENC_IN_SCALE),
        -128,
        127
    ).astype(np.int8)

    if input_int8.shape != (TARGET_H, TARGET_W, 3):
        raise ValueError(f"shape mismatch: {input_int8.shape}")

    # 5. 保存 HWC int8，供 PYNQ Encoder 直接读取
    save_path = os.path.join(SAVE_DIR, f + '.npy')
    np.save(save_path, input_int8)

    if i < 5 or (i + 1) % 20 == 0 or (i + 1) == len(files):
        print(
            f"[{i+1:04d}/{len(files):04d}] {f} -> {os.path.basename(save_path)} "
            f"shape={input_int8.shape}, "
            f"range=[{int(input_int8.min())},{int(input_int8.max())}], "
            f"mean={float(input_int8.mean()):.3f}, std={float(input_int8.std()):.3f}",
            flush=True
        )

print("=" * 80)
print("✅ 处理完成")
print("数据保存在:", SAVE_DIR)
print("=" * 80)
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import math

# --- 配置 ---
SAVE_DIR = '/home/xilinx/jupyter_notebooks/duxu/pynq_vqvae/imgs_preprocessed_1920x1080'
ENC_IN_SCALE = 0.015625  # 必须与预处理时的 scale 一致 (2^-6)

# 1. 获取所有 npy 文件
npy_files = sorted([f for f in os.listdir(SAVE_DIR) if f.endswith('.npy')])
num_imgs = len(npy_files)

if num_imgs == 0:
    print(f"❌ 错误：在 {SAVE_DIR} 文件夹下未找到 .npy 文件。")
else:
    print(f"🔍 找到 {num_imgs} 张预处理图片，正在准备预览...")

    # 2. 计算展示网格的行数和列数 (假设每行显示 5 张)
    cols = 5
    rows = math.ceil(num_imgs / cols)

    fig, axes = plt.subplots(rows, cols, figsize=(20, 4 * rows))
    axes = axes.flatten() # 展平方便循环

    for i in range(len(axes)):
        if i < num_imgs:
            # 加载 int8 数据
            data_int8 = np.load(os.path.join(SAVE_DIR, npy_files[i]))
            
            # 还原量化与归一化
            # int8 -> fp32 -> 原图范围 [0, 1]
            data_fp32 = data_int8.astype(np.float32) * ENC_IN_SCALE
            img_rgb = (data_fp32 * 0.5 + 0.5).clip(0, 1)
            
            # 显示图片
            axes[i].imshow(img_rgb)
            axes[i].set_title(f"Img: {i}\n{npy_files[i][:20]}...", fontsize=8)
            axes[i].axis('off')
        else:
            # 隐藏多余的子图格
            axes[i].axis('off')

    plt.tight_layout()
    plt.show()
    print("✅ 所有图片展示完毕。请检查图片是否存在偏色、过暗或噪点。")
```
